```python
# ============================================================
# BƯỚC 1: INTERNET ON - TẢI WHEELS VÀ MODEL YOLO11s, YOLO11m
# ============================================================
from pathlib import Path
import shutil
import subprocess
import sys

WHEEL_DIR = Path("/kaggle/working/wheels")
WHEEL_DIR.mkdir(parents=True, exist_ok=True)

# 1. Khai báo packages (Đã thêm sahi)
packages = [
    "ultralytics==8.4.121",
    "pycocotools==2.0.11",
    "pandas==3.0.5",
    "matplotlib==3.11.1",
    "seaborn==0.13.2",
    "tqdm==4.67.3",
    "sahi" # Thêm SAHI vào đây
]

# 2. Cài đặt vào môi trường hiện tại
subprocess.run([sys.executable, "-m", "pip", "install", "-q", *packages], check=True)
print("✅ Packages installed.")

# 3. Tải offline wheels (Bao gồm dependencies)
subprocess.run([sys.executable, "-m", "pip", "download", *packages, "-d", str(WHEEL_DIR)], check=True)
print("✅ Offline wheels downloaded.")

# 4. Tải model yolo11s và yolo11m
from ultralytics import YOLO

models_to_download = ["yolo11s.pt", "yolo11m.pt"]
for model_name in models_to_download:
    model_path = Path(f"/kaggle/working/{model_name}")
    # Load model để trigger việc download
    _ = YOLO(model_name)
    src_path = Path(model_name).resolve()
    dst_path = model_path.resolve()
    
    if src_path.exists() and src_path != dst_path:
        shutil.copy2(src_path, dst_path)
    elif not dst_path.exists():
        cache_cands = list(Path.home().rglob(model_name))
        if cache_cands:
            shutil.copy2(cache_cands[0], dst_path)
    print(f"✅ Model downloaded: {model_path}")

# 5. Zip thư mục wheels
zip_path = shutil.make_archive("/kaggle/working/visdrone_yolo_wheels", "zip", WHEEL_DIR)

print("\n================================")
print("✅ PREPARATION COMPLETE")
print("================================")
print(f"Wheel ZIP : {zip_path}")
```
